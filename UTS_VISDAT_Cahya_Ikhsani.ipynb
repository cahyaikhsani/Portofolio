```python
import matplotlib.pyplot as plt

bulan = ['Januari', 'Februari', 'Maret', 'April', 'Mei', 'Juni']
pendapatan = [120, 135, 125, 150, 180, 210]

plt.title("Tren Pendapatan Bulanan (Januari - Juni)", fontsize=14, fontweight="bold")
plt.plot(bulan, pendapatan, color='purple')
plt.xlabel("Bulan")
plt.ylabel("Pendapatan (Juta Rp)")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

warna_soft = sns.color_palette('pastel')
nama_kategori = ['Elektronik', 'Pakaian', 'Alat Tulis', 'Peralatan Rumah', 'Kosmetik']
nilai_angka = [450, 820, 310, 540, 670]

plt.title("Perbandingan Total Penjualan per Kategori Produk", fontsize=14, fontweight="bold")
plt.bar(nama_kategori, nilai_angka, color=warna_soft)
plt.xlabel("Kategori Produk")
plt.ylabel("Total Terjual (Unit)")
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

bulan = ['Januari', 'Februari', 'Maret', 'April', 'Mei', 'Juni']
biaya_iklan = [10, 12, 11, 15, 20, 25]
pendapatan =[120, 135, 125, 150, 180, 210]

r = np.corrcoef(biaya_iklan, pendapatan)[0, 1]
print(f"Koefisien korelasi (r) = {r:.4f}")

for i, nama in enumerate(bulan):
    plt.annotate(nama, (biaya_iklan[i], pendapatan[i]),
                 textcoords="offset points", xytext=(0, 8), ha='center', color='black')

plt.scatter(biaya_iklan, pendapatan, color='blue')
plt.xlabel("Biaya Iklan (Juta Rp)")
plt.ylabel("Pendapatan (Juta Rp)")
plt.title("Perbandingan Biaya Iklan dan Pendapatan", fontsize=14, fontweight="bold")
plt.show()
```
