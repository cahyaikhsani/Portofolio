```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram, set_link_color_palette

data = {
    'Nama': ['Andi', 'Slamet', 'Citra', 'Dewi', 'Bejo'],
    'Frekuensi_Belanja': [1, 2, 8, 9 ,5],
    'Total_Belanja': [2, 3 ,15, 16, 9]
}
df = pd.DataFrame(data)
X = df[['Frekuensi_Belanja', 'Total_Belanja']]

linked = linkage(X, method='ward')

set_link_color_palette(['#1f77b4', '#2ca02c', '#ff7f0e'])
plt.figure(figsize=(8, 5))
dendrogram(
    linked,
    labels=df['Nama'].values,
    leaf_font_size=12,
    color_threshold=10,
    above_threshold_color='#7f7f7f'
)
plt.axhline(y=10, color='red', linestyle='--', label='Garis_Potong (2 Klaster)')

plt.title('Dendogram Segmentasi 5 Pelanggan', fontsize=14, fontweight='bold')
plt.xlabel('Nama Pelanggan', fontsize=12)
plt.ylabel('Jarak Ketidakmiripan', fontsize=12)
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import plotly.express as px

data = {
    'Kategori': [
        'Elektronik', 'Elektronik', 'Elektronik', 'Elektronik',
        'Fashion', 'Fashion', 'Fashion',
        'Makanan', 'Makanan'
    ],
    'Sub_Kategori': [
        'Laptop', 'Laptop', 'Smartphone', 'Smartphone',
        'Sepatu', 'Sepatu', 'Pakaian',
        'Minuman', 'Camilan'
    ],
    'Produk': [
        'MacBook Air', 'Asus ROG', 'iPhone 15', ' Samsung S24',
        'Sneakers Pria', 'Sepatu Lari', ' Jaket Denim',
        'Kopi Arabika', 'Keripik Singkong'
    ],
    'Omzet_Juta': [45, 35, 50, 30, 25, 20, 15, 12, 8]
}
df = pd.DataFrame(data)
warna_kustom = {
    '(?)': '#ECEFF1',
    'Elektronik': '#1f77b4',
    'Fashion': '#ff7f0e',
    'Makanan': '#2ca02c'
}
fig = px.treemap(
    df,
    path=[px.Constant("Total Penjualan Toko"), 'Kategori', 'Sub_Kategori', 'Produk'],
    values='Omzet_Juta',
    color='Kategori',
    color_discrete_map=warna_kustom,
    title='Treemap Penjualan'
)

fig.update_traces(
    textinfo="label+value+percent parent",
    textfont_size=13
)

fig.update_layout(margin=dict(t=50, l=20, r=20, b=20))
fig.show()
```
